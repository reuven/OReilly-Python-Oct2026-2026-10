# What are functions?

We've seen a bunch of functions (and methods) from the first day of this course:

- `print` (displays things on the screen)
- `input` (gets inputs from the user)
- `len` (returns the length of a string, list, tuple, or dict)
- `sum` (takes a list of numbers and returns their total)
- `type` (returns the type of value we hand it)

We don't need functions! Or we don't need more functions than Python comes with. We could make do with just the functions (and methods) that Python has, out of the box. And many of those are probably not absolutely necessary.

Why, then, do we have so many functions? Why are we going to talk today about writing new functions?

Abstraction. The idea of abstraction is that we can ignore the details and think at a higher level.

Writing functions allows us to go up a level (or more) of abstraction. We put the details in the function's definition, and once we know that the function works, we never need to think about those details again. Instead, we can just invoke the function, enjoy what it does, and ignore the internals. If and when we need to fix those internals, then we can open the function up -- but until then, we can think at our higher level.

Once I have a number of low-level functions defined, then I can work at a higher level. Much of programming is writing low-level functions so that we can think at a higher level, and never worrry about the low-level functions' internals again.

At the end of the day, a functions allows us to put a name to a bunch of functionality. It's basically defining a verb. A new verb that Python will know after we've defined it, and which will execute whatever steps we need.

So no... we don't need to be able to define functions... but it makes writing software *so* much easier and more straightforward.



# Let's define a function!

To define a new function, we need a few things:

1. The keyword `def`, which means "define"
2. Immediately after `def`, we put the name of the function we want to define. This is really a variable assignment! The variable will be assigned a function value.
3. After the function name, we put `()`. Right now, these will be empty, but pretty soon we'll put parameter (variable) names inside of them.
4. At the end of the line, we put `:`. Why? Because the function body will start on the following line, and it will be indented.
5. The function body follows -- it can contain any code we want, at all. It's usually a good idea to keep the function body relatively short, but sometimes that isn't possible.

```python
def hello():             # function definition line, with "def", empty () for no parameters, and then :
    print('Hello!')      # one-line function body. Here, we're invoking "print", and asking it to print "Hello!"
```

Defining a function doesn't execute it. If you want to execute a function, you need to actually mention its name and then use `()`.

```python
hello()   # When Python see (), it looks to its left and asks -- is this thing "callable" (i.e., executable)? If so, it runs
```

```python
callable(hello)    # pass a function as an argument to another function!
```

```python
# what happens if I don't use ()?

hello    # we got back the function definition, or the function "object," but we didn't execute the function itself
```

1. Define a function, `calc`, that when invoked asks the user three questions:
    - What is the first number?
    - What is the operator?
    - What is the second number?
2. Assign the responses to `first`, `op`, and `second`, in order.
3. If `first` and `second` can be turned into integers, do so.
    - If not, then print an error message of some sort.
5. If the operator is either `+` or `-`, then calculate the result, and assign to `result`
    - If it's neither of these operators, then `result` can be set to `'bad operator'`
5. If the values are OK and the operator is OK, then print the entire expression plus its result.

Example:

    calc()
    First: 10
    Operator: +
    Second: 8
    10 + 8 = 18



```python
def calc():
    first = input('First: ').strip()
    op = input('Operator: ').strip()
    second = input('Second: ').strip()

    if first.isdecimal() and second.isdecimal():
        first = int(first)    # it's guaranteed that we can do this here, because we know (from line 6) that the conversion will work
        second = int(second)

        if op == '+':
            result = first + second
        elif op == '-':
            result = first - second
        else:
            result = '(bad operator)'

        print(f'{first} {op} {second} = {result}')

    else:
        print(f'Both {first} and {second} need to be numeric; try again.')
```

# This is annoying!

This function, `calc`, works -- but it requires that we be in front of our computer every time we want to run this function. It's very inconvenient for this to be the case.

It makes more sense for us to invoke the function, passing some values to it. Then we don't need to be pestered with the `input` function.

How would that look?

When we define our function, we put `()` after the name. We can, instead, put one or more variable names inside of the `()`. If we do that, then we're saying two things: (1) you must pass arguments to invoke this function, and (2) the arguments will be assigned to the parameters.

```python
def hello(name):
    print(f'Hello, {name}!')
```

```python
# parameters:  name
# arguments:  'Reuven'   # assignment of the first argument to the first parameter

hello('Reuven')
```

```python
hello('someone else')
```

```python
# what happens if I now invoke hello without any arguments?

hello()
```

```python
hello('a', 'b')
```

- The first `def hello` defined a function that take zero arguments.
- The second `def hello` defined a function that takes one argument.
- WHen we invoke `hello` after both definitions, only the second one is still around. The first is gone.

# Return values

When we invoke `input`, we can put it on the right side of assignment (`=`). That's only possible because it *returns* a value. That is, `input` gives us something back -- in mathematical terms, we can say that the call to `input` is replaced by a value.

`len` also returns a value, an integer.

What about other functions? What do they return? Each can return whatever it wants. `str.split`, for example, returns a list of strings.

And our functions, the ones we have defined so far today? They all return the special value `None`, because if you don't explicitly return a value in Python, the function returns `None`.

Some programming languages distinguish between function (which return values) and procedures (which don't). Python doesn't make this distinction, because *every* function returns something -- even if it's `None`.

In order to return a value, the function uses the keyword `return`:

- If there isn't any `return`, the function returns `None`
- If there is a `return`, but it doesn't have a value, the function returns `None`.
- Otherwise, whatever value is to the right of `return` is returned to the caller.

```python
def hello(name):
    return f'Hello, {name}!'  # notice that return isn't a function -- so no () after it
```

```python
s = hello('Reuven')   # notice -- the call to hello is on the right side of assignment!
```

```python
print(f'>>> {s} <<<')
```

It's a far better idea for a function to return a value than to print it:

- If you `print` from your function, then the value cannot be retrieved/stored/used.
- If you `print` from your function, then the user doesn't have control over what is displayed and what is stored.

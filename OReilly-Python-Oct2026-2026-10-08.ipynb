# Loops

Let's say that I have a string, `'abcd'`, and I want to print each of the characters in the string. I can do that:

```python
s = 'abcd'  # assign the string to a variable

print(s[0])
print(s[1])
print(s[2])
print(s[3])
```

# What's the problem?

We're repeating ourselves. The same code (more or less) is taking place on lines 3, 4, 5, and 6.

We generally want to avoid such repetition. In fact, there's a famous rule, DRY -- don't repeat yourself!

What's wrong with repeating yourself?

- More work writing it
- More work reading it -- just reading code is hard, and reading repeated code is extra-hard
- More work maintaining it -- if we find a bug in the code, then we have to fix it wherever that code appears. That can be a one-time thing if the code is centralized/not repeated, or it can be a day-long event, tracking down where the code exists.

One of the best ways to "DRY up" your code is to use a *loop*, where code repeats a number of times. Each repetition, known as an *iteration*, can be slightly differnt from the others.

In Python, we have exactly two types of loops:

- `for` loops, more common and go through every element of a string or similar sequence
- `while` loops, which are less common, and are very similar to `if` statements that repeat.



```python
# simple loop to do what we did above

s = 'abcd'

for one_character in s:
    print(one_character)
```

# Syntax of a `for` loop

1. We start with the keyword `for`
2. Then we have the *loop variable*, in this case, `one_character`. This is where the loop assigns the new value with each iteration.
3. Then we have the keyword `in`
4. Then we have the value over which we want to iterate. Here, it'll be `s`, the variable, which refers to a string.
5. At the end of the line, we have a `:`, indicating that the next line needs to be indented.
6. The next line starts an indented block. Its execution will continue until the indentation ends (or until the program ends)
7. The loop will execute once for each character in `s`, until we get to the end. At that point, the loop exits silently and continues with the rest of the program.

# Conversation of a `for` loop

- `for` turns to the value at the end of the line (`s`, in this case) and asks: Are you iterable? Meaning, do you know how to behave in a `for` loop?
    - If not, then the loop exits with an error
- If the value is iterable, then `for` asks for its next value
    - If we've reached the end of the values, then the loop exits and we continue with the code after the loop.
- The "next value" is assigned to the loop variable (in this case, `one_character`.
- Then the body of the loop executes with the defined loop variable.
- When we get to the end of the loop body, we go back and ask (again) for the next value. It'll either be assigned to the loop variable, or the loop will be told that there are no more values, and we should exit

# Notice a few things here

- The name of the loop variable has *NOTHING* to do with what values we get.
- We don't know in advance how many iterations there will be, or what values we'll get


# Exercise: Vowels, digits, and others

We're going to ask the user to enter a text string. We'll then iterate over that string, one character at a time, and check:

- Is it a vowel? (a, e, i, o. u)
- Is it a digit? (0-9)
- Is it something else?

To do:

1. Define three variables, `vowels`, `digits`, and `others`, and set them to all be 0.
2. Ask the user, with `input`, to enter some text.
3. Use a `for` loop to go through the text, one character at a time.
4. Print each of the counters.

If I enter:

    Text: hello!! 123
    vowels: 2
    digits: 3
    others: 6

```python
vowels = 0
digits = 0
others = 0

text = input('Enter text: ').strip()   # assign the user's input to "text", removing leading/trailing whitespace

for one_character in text:
    if one_character in 'aeiou':    # is one_character a vowel?
        vowels += 1                 # ... if so, then add 1 to vowels
    elif one_character.isdigit():   # is one_character a digit, 0-9?
        digits += 1                 # ... if so, then add 1 to digits
    else:
        others += 1                 # neither a vowel nor a digit? Add 1 to others

print(f'vowels: {vowels}')
print(f'digits: {digits}')
print(f'others: {others}')
```

# What if I just want to repeat something?

Let's say that I'm in a great mood, and want to celebrate via code:

```python
print('Hooray!')
print('Hooray!')
print('Hooray!')
```

```python
# the above code works... but it's not very DRY. how can we get this to not repeat?

# what if I iterate over the number 3?

for one_item in 3:
    print('Hooray!')
```

# The solution: `range`

We can invoke `range(n)`, which is designed to be used in a `for` loop:



```python
for one_item in range(5):   # range(5) iterates 5 times
    print('Hooray!')
```

# What is `range`? What are we getting with each iteration?

```python
for one_item in range(5):   # range(5) iterates 5 times
    print(f'[{one_item}] Hooray!')
```

# Iterating over `range`

When we iterate over `range`, we get integers:

- Starting with the initial number (default is 0)
- Up to and *not* including the final number (which is the only mandatory argument for range)

`range(n)` will give you, on each iteration, the next number, until it reaches `n-1`, which is the final value it returns.

```python
# Back to vowels, digits, others briefly:

vowels = 0
digits = 0
others = 0

text = input('Enter text: ').strip()   # assign the user's input to "text", removing leading/trailing whitespace

for one_character in text:
    if one_character in ['a', 'e', 'i', 'o', 'u']:    # this still works! But... what's the difference?
        vowels += 1                 # ... if so, then add 1 to vowels
    elif one_character.isdigit():   # is one_character a digit, 0-9?
        digits += 1                 # ... if so, then add 1 to digits
    else:
        others += 1                 # neither a vowel nor a digit? Add 1 to others

print(f'vowels: {vowels}')
print(f'digits: {digits}')
print(f'others: {others}')
```

# What's the difference?

If you define `'aeiou'`, then you have defined a string.

If you define `['a', 'e', 'i', 'o', 'u']`, then you have defined a *list*, a completely different data type, but one that has a lot of ties to strings and a lot of overlapping functionality.

It's a bit faster and smaller to use a string in this case. But we'll talk more about lists soon.

```python
for counter in range(5):
    print(counter)
```

```python
for counter in range(100):
    print(counter, end=' ')
```

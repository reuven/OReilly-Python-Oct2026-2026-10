# Loops

Let's say that I have a string, `'abcd'`, and I want to print each of the characters in the string. I can do that:

```python
s = 'abcd'  # assign the string to a variable

print(s[0])
print(s[1])
print(s[2])
print(s[3])
```

# What's the problem?

We're repeating ourselves. The same code (more or less) is taking place on lines 3, 4, 5, and 6.

We generally want to avoid such repetition. In fact, there's a famous rule, DRY -- don't repeat yourself!

What's wrong with repeating yourself?

- More work writing it
- More work reading it -- just reading code is hard, and reading repeated code is extra-hard
- More work maintaining it -- if we find a bug in the code, then we have to fix it wherever that code appears. That can be a one-time thing if the code is centralized/not repeated, or it can be a day-long event, tracking down where the code exists.

One of the best ways to "DRY up" your code is to use a *loop*, where code repeats a number of times. Each repetition, known as an *iteration*, can be slightly differnt from the others.

In Python, we have exactly two types of loops:

- `for` loops, more common and go through every element of a string or similar sequence
- `while` loops, which are less common, and are very similar to `if` statements that repeat.



```python
# simple loop to do what we did above

s = 'abcd'

for one_character in s:
    print(one_character)
```

# Syntax of a `for` loop

1. We start with the keyword `for`
2. Then we have the *loop variable*, in this case, `one_character`. This is where the loop assigns the new value with each iteration.
3. Then we have the keyword `in`
4. Then we have the value over which we want to iterate. Here, it'll be `s`, the variable, which refers to a string.
5. At the end of the line, we have a `:`, indicating that the next line needs to be indented.
6. The next line starts an indented block. Its execution will continue until the indentation ends (or until the program ends)
7. The loop will execute once for each character in `s`, until we get to the end. At that point, the loop exits silently and continues with the rest of the program.

# Conversation of a `for` loop

- `for` turns to the value at the end of the line (`s`, in this case) and asks: Are you iterable? Meaning, do you know how to behave in a `for` loop?
    - If not, then the loop exits with an error
- If the value is iterable, then `for` asks for its next value
    - If we've reached the end of the values, then the loop exits and we continue with the code after the loop.
- The "next value" is assigned to the loop variable (in this case, `one_character`.
- Then the body of the loop executes with the defined loop variable.
- When we get to the end of the loop body, we go back and ask (again) for the next value. It'll either be assigned to the loop variable, or the loop will be told that there are no more values, and we should exit

# Notice a few things here

- The name of the loop variable has *NOTHING* to do with what values we get.
- We don't know in advance how many iterations there will be, or what values we'll get

- Is it a vowel? (a, e, i, o. u)

# Iterating over `range`

When we iterate over `range`, we get integers:

- Starting with the initial number (default is 0)
- Up to and *not* including the final number (which is the only mandatory argument for range)

`range(n)` will give you, on each iteration, the next number, until it reaches `n-1`, which is the final value it returns.

```python
# Back to vowels, digits, others briefly:

vowels = 0
digits = 0
others = 0

text = input('Enter text: ').strip()   # assign the user's input to "text", removing leading/trailing whitespace

for one_character in text:
    if one_character in ['a', 'e', 'i', 'o', 'u']:    # this still works! But... what's the difference?
        vowels += 1                 # ... if so, then add 1 to vowels
    elif one_character.isdigit():   # is one_character a digit, 0-9?
        digits += 1                 # ... if so, then add 1 to digits
    else:
        others += 1                 # neither a vowel nor a digit? Add 1 to others

print(f'vowels: {vowels}')
print(f'digits: {digits}')
print(f'others: {others}')
```

# Indexes

In C (and similar languages), a `for` loop has three parts:

- Starting integer (usually 0)
- An action to do with each iteration (e.g., `i = i + 1`)
- Ending condition (e.g., `i < 5`)

As you can see, a `for` loop in such a language is all about the numbers. You can go down, you can go up, you can go up by large increments, but it's all about the numbers.

You then use the index to retrieve whatever character you want from the string. There's no such thing as iterating over the characters in a string.

In Python, we say: Why, if we want to iterate over the characters of a string, do we need a separate data structure (i.e., an integer)? Just give me the characters, one at a time!

That's why our `for` loop looks and works the way it does.

However... there are times when you'll want to use the index, either for calculations or for displaying the index next to each character. How can we do that?


```python
# Option 1: Pretend we're in C
# warning: very ugly code ahead

s = 'abcd'

for i in range(len(s)):    # get the length of s, then run range on that
    print(f'[{i}] {s[i]}')
```

```python
# Option 2: Set the index manually
# here, I'm getting each character -- and as each character comes to us from the loop, we add 1 to the index

i = 0

for one_character in s:  
    print(f'[{i}] {one_character}')
    i += 1    # this adds 1 to i
```

```python
# Option 3: Use the "enumerate" function that comes with Python
# enumerate gets, as input, the string (or other iterable you want to get the index of)

for i, one_character in enumerate(s):     # yes, we're iterating over 2 values with each iteration! We have 2 loop variables!
    print(f'[{i}] {one_character}')
```

# Exercise: Powers of 10

If I have a decimal number, such as 3579, then I can rewrite it as the sum of powers of 10:

- 3 * 10**3
- 5 * 10**2
- 7 * 10**1
- 9 * 10**0  # anything to the 0 power is 1

I want you to ask the user to enter a number. (Assume that they do.) Print each of the digits in the same format as I just did in my example.

```python
number = input('Enter a number: ').strip()

if number.isdigit():
    for index, one_digit in enumerate(number):   # iterate over the number (string) and get its index
        exponent = len(number) - index - 1       # calculate the exponent -- length - index - 1
        print(f'{one_digit} * 10**{exponent}')   # print things using an f-string

else:
    print(f'{number} is not numeric; try again.')
```

# Controlling our loops

So far, we've seen that we can iterate over all of the characters in a string with a `for` loop. What if we want to iterate over only the first half, exiting when the index is before the end?

Right now, we have two ways to do that:

1. We can iterate over a slice (i.e., a partial string)
2. We can get the index, and only print if the index is low

Neither of these really gets to the heart of the problem: Sometimes, when we're iterating in a `for` loop, we discover that we can/should go onto the nextr iteration, or we should exit the loop entirely.

When would these be appropriate? Often:

- We need to exit the loop immediately, if we have accomplished our goal -- this is done with `break`
- We might not want to exit the loop, but we might want to exit the current iteration, if it's clearly unnecessary. If the current input from the user is completely invalid, then we should move onto the next one, not go through the entire loop body. This is done with `continue`

```python
# end the loop with break

s = 'abcde'
look_for = 'c'

print('Before')
for one_character in s:
    if one_character == look_for:
        break   # end the loop right here!

    print(one_character)
print('After')    
```

```python
# end the iteration (but not the loop) with continue

s = 'abcde'
look_for = 'c'

print('Before')
for one_character in s:
    if one_character == look_for:
        continue   # end this iteration, but keep going through the rest of the loop

    print(one_character)
print('After')    
```

# Exercise: Sum digits

1. Ask the user to enter a string with digits.
2. Set `total` to 0.
3. Go through each character in the user's input string.
    - If it's a `.`, then stop the loop entirely, and print `total`
    - If it's a non-digit (other than `.`), then scold the user and go on
    - If it's a digit, then add to `total`.
4. Print `total`.

```python
total = 0
digits = input('Enter some digits: ').strip()

for one_character in digits:
    if one_character == '.':
        print(f'Ending; total is {total}')
        break    # end the loop!

    if one_character.isdigit():
        total += int(one_character)
        print(f'Added {one_character}; total is now {total}')

    else:
        print(f'{one_character} is not a digit; ignoring.')
        continue
```

```python
total = 0
digits = input('Enter some digits: ').strip()

for one_character in digits:
    if one_character == '.':
        print(f'Ending; total is {total}')
        break    # end the loop!

    if not one_character.isdigit():
        print(f'{one_character} is not a digit; ignoring.')
        continue
        
    total += int(one_character)
    print(f'Added {one_character}; total is now {total}')
```

# `while` loops

In a `for` loop, we start at the beginning, and go through each element of the sequence. We either get to the end or invoke `break` to stop. But the idea in a `for` loop is that we're iterating over each element, one at a time. We stop at the end.

By contrast. `while` loops are meant for situations when we don't know how many iterations we'll need. Maybe we'll need many, and maybe not that many. Maybe none at all.

A `while` loop is basically an `if` statement whose body repeats so long as the condition is `True`.

- An `if` looks to its right, and evaluates the value there. If the value is `True`, then the block runs.
- In the case of `while`, the same thing happens. The only difference is that after the block runs, we go *back* to the `while`, and it runs again.

```python
# example while loop

x = 5

while x > 0:
    print(x)
    x -= 1   # x = x - 1
```

Note that both `break` and `continue` work just fine in both `while` and `for` loops.

# Tuples

Tuples are the third member of the "sequence" family, along with strings and lists.

- Strings contain characters, and are immutable
- Lists contain anything, and are mutable
-

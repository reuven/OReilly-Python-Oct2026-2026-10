# What are modules?

We've discussed the DRY ("don't repeat yourself") rule for programming:

- If we have roughly the same code, several lines in a row, then we can replace them with a loop.
- If we have roughly the same functionality in several places in a program, we can write a function, and then invoke it wherever we want
- if we have roughly the same functionality in several different programs, we can store the functions and/or data in a *library*.

In Python, we call our libraries "modules." (A module is a single file containing Python code; if you have more than one file in your library, we in the Python world call it a "package.")

Modules in Python give us at least three types of functionality:

1. They contain the data and the functions we might want to use.
2. They provide us with *namespaces*, separations between different sets of defined names.
3. Modules (and packages) allow us to distribute our code to other people, working on other programs.

Modules are among the most important and frequently used pieces of a Python program. They're complicated under the hood, but using them is very easy, and writing simple modules is also very easy.

# Let's use a module!

If you want to use a module, you use the `import` statement. `import` takes the name of a module as its argument, uses that to load a module from disk (from the file where it's defined), creates a module in Python's memory, and assigns that module to a variable.

```python
import random
```

# What do we get from the above line?

1. `import` is not a function. Don't invoke it with `()`. Rather, just say `import` and then after a space, put the name of the module you want to load and then define.
2. The name that we put after `import` is *not* a string, and not a filename. Rather, it's the name of the variable we want to create, *plus* the name of the file we want to load. In this case, we're telling Python to load `random.py`, where `py` is the suffix for most Python programs and all Python modules.
3. Many other languages have similar functionality to `import`, but there you often pass a string or a filename, and that's what is loaded. That's not how it works here.

After our `import` has run, the `random` variable is defined, and it contains a module called `random`, with all of the data and functionality that were defined in that module.

```python
type(random)  # what kind of object, or value, do we have in the "random" variable?
```

```python
random   # show me the module's printed representation
```

# Using names from modules

The module itself is just a warehouse for names. If you want to retrieve a value from a module, just name module, then `.`, then the name you want:

This retrieves the value. But if the value is a function, you'll still need to provide `()` and (potentially) one or more arguments.

If it's just data, then you can say

    random.NAME

1. Generate a random `secret` from the `random` module, between 0 and 100.

```python
import random

counter = 0
secret = random.randint(0, 100)

while True:
    guess_text = input('Guess: ').strip()

    if guess_text.isdigit():
        counter += 1
        guess = int(guess_text)
        if guess == secret:
            print(f'You got it in {counter} guesses.')
            break

        elif guess < secret:
            print('Too low!')

        else:
            print('Too high!')

    else:
        print(f'{guess_text} is not numeric; try again')
```

# What does `from .. import` do?

1. It loads the entire module into memory.
2. But it only exposes and defines the one that you specify.

Many *many* people believe that `from .. import` saves memory, because it ony imports a handful of files. That's not true -- when you say `from .. import`, Python needs to find, then define, just one name -- to do that, it needs all of the names, and the entire module, loaded.

# Variations on `import`

- `import MODNAME` -- defines the module variable
- `from MODNAME import NAME` -- defines just the names that we want to import
- `import MODNAME as ALIAS` -- loads the module, but changes the name of the variable it defines
- `from MODNAME import NAME as ALIAS` -- load the name ou want, and immediately change the name

# Names and values in Python

When I assign a value to a variable:

    x = 5

I'm saying that the name `x` refers to the value `5`. 

When we define a function:

    def hello():
        return 'Hello!'

We're defining a function and then assigning that function value to the name `hello`.

If I import a module:

    import random

Python is creating the module object/value, and then assigning it to the variable `random`.

In all three cases, we're assigning to a name ("identifier"), and Python doesn't really care whether it's an integer, string, dict, function, or module. In all cases, they're just values.

Many languages have one set of names for functions and another for data. That's not the case in Python -- any identifier can refer to any value, be it a function, module, or something else.

# What happens when you import a module?

It might seem very clear at first what happens when you use `import`:

- Python finds the file
- Python loads the file
- You now have the data from the file, as a module

It's more interesting and more complex than that -- we're not going to go into all of the details, but we should review some of them, because they do affect how your program works. 

Finding the file is easy -- we already discussed `sys.path`. Python does a `for` loop on `sys.path` -- if you asked for `mymod`, it looks for `mymod.py` in each directory in `sys.path`. The important thing is: The first match wins. So if you put a module, e.g., `xml.py` in the current directory, you have now lost access to the `xml.py` that comes with Python! Because the current directory has priority over all directories in `sys.path`.

What does it mean for Python to "load" the file? For many people, it means that the definitions are somehow loaded into memory, but aren't actually defined until we need them.

This isn't true! It's far easier than this.

When you `import` a module, the module file is read and executed, line by line, immediately when you say `import MODULE`. If your module is very large, or takes a long time to load, then the program that imports it will be waiting on that line for a little while. 

When the module is executed, it's done using a mechanism in Python that allows all of the definitions to be attached to a module. So the global variables defined in a module are turned into attributes on the module object.

- global `x` -> `mymod.x`
- global `y` -> `mymod.y`
- global `hello` -> `mymod.hello`

This seems to mean that we can put any code we want in a module, and it'll be executed when the module is imported! Can that be true?

# `__name__` is very special

It always reflects the current *namespace* so you'll know if the module is running as a result of `import` or if it's running as a standalone. 

The way it makes that distinction is with `__name__`:

- If the module was imported, then `__name__` is a string, the name of the module
- If the module is being run as a fullf-ledged program, then `__name__` is still a string, but is `'__main__'`.

How does this help us?

We can use an `if` statement in our module, and only print certain things when we're developing, not in production.

# Exercise: Using `glob` from the standard library

1. The `glob` module allows you to find files matching a pattern, using `*` and `?`.
2. Write a program that asks the user to enter a pattern.
3. Find all filenames matching that pattern.
4. For each filename, print the size of the file alongside the name. (Assume that the file contains only text, and you can then iterate over each line, adding to the total length.)

```python
import glob

pattern = input('Enter pattern: ').strip()

for one_filename in glob.glob(pattern):
    size = 0
    for one_line in open(one_filename):
        size += len(one_line)

    print(f'{one_filename} : {size}')
```

# Two scenarios for `import` and the whole `if __name__` stuff

- We have a module, `menu.py`. It contains two parts:
    - `def menu`, a function definition
    - `if __name__ == '__main__'`, followed by `print` in the block

### Scenario 1:
- From another Python program, we say `import menu`
- Because a separate program is saying `import menu`, Python sets the `__name__` variable to be the string `'menu'` inside of `menu.py`.
- It executes `menu.py` in that context (i.e., `__name__` is now `'menu'`
- The first part of `menu.py` executes, because it's a `def` and has no guards on it
- The second part of `menu.py` does *NOT* execute, because it checks that `__name__ == '__main__'`, and that is *NOT* true here.
- The end result is that the `menu` function is defined and available to the caller, but nothing is printed or asked of the user.

### Scenario 2:
- From the terminal, we invoke `menu.py` as a standalone program.
- Because we're invoking it as a standalon eprogram, Python sets the `__name__` variable to be the special string `'__main__'`
- It executes `menu.py` in that context (i.e., `__name__` is now `'__main__'`)
- The first part of `menu.py` executes, because it's a `def` and has no guards on it
- The second part of `menu.py` *DOES* execute, because it checks if `__name__ == '__main__'`, which IS TRUE NOW, because we're running the program standalone, i.e., not importing it
- The end result is not just that `menu` defines `menu.menu`, but it also shows with a brief demo the `menu` function in adction.

# PyPI

The Python standard library is big, but it's still very small compared with the universe of things that people have created in the Python world. 

Any module that is distributed with Python is in a bind -- it's very popular, and easy to use, but authors cannot update whenever they want. For this and some other reasons, we needed a repository for third-party packages in Pytohn. That's PyPI.

Python Package Index contains many *many* packages that you can download and install.

# Downloading and installing

For years, the standard way to install packages from PyPI was a program called `pip` (Python installs package, or the like). You could say `pip install ABCD`, and it would look on PyPI for `ABCD`, and if it was there, it would download it and install it on your computer.

But... there were some issues. Let's say you're working on project A which needs mixing 1.0. You're also working on project B which needs mixing 2.0. Let's assume that each version of `mixing` is totally incompatible with others.

The traditional response is to use `venv`, but those are annoying.

`pip` knows how to download into a venv, which keeps different versions separate. But people were getting frustrated. 

`uv` is the new darling of the Python world. You can use it to install packages, and to upload new packages to PyPI, too!

If you're using `pip` (the default), then installing something from PyPI is as easy as 

    pip install PACKAGENAME

If you're using uv, then you can say

    uv add PACKAGENAME

# Dictionaries (`dict`)

We've seen in the last two weeks a few different data structures:

- Numbers (`int` and `float`)
- Text (`str`)
- Container sequences (`list` and `tuple`)

In the case of strings, lists, and tuples, the values are identified with an index that starts at 0 and goes up to the `len` - 1. This works, but it means that the index has no relationship with our data. A value could be at index 2, or 10, or 158. 

But what if we have workers at our company with ID numbers? Then we'll (maybe) store a tuple about each worker at a different element in a list, but that element number and the ID number will have nothing to do with one another.

You can think of a dict as kind of like a list, except that you can decide what the index is (we call it a "key" in dicts) for each value. That is: You can use any number you want, not just 0, 1, 2, 3, etc. You can even use strings or tuples as your keys.

Better yet, searching and retrieving from a dict is extremely efficient and fast -- so if you're going to search for a value in a list, you might (sometimes, often?) be better off using a dict.

# Dicts aren't only in Python

Other languages call our "dicts" by other names:

- Hash tables
- Hash maps
- Maps
- Hashes
- Key-value stores
- Name-value stores
- Associative arrays

The basic idea, no matter the name and no matter the language is:

- You are storing pairs of information
- The pair has a *key* and it has a *value*
- The key is what we use to store, retrieve, and search
- The value is ultimately what's interesting/important for us

# To create a dict

In Python, we use `{}` for our dictionaries:

- `{}` by themselves are the "empty dict," similar to the empty string, empty list, and empty tuple
- Use `{}` with one or more pairs for a non-empty dict
- Each key is separated from its value with `:`, e.g., `'a':10`
- Each pair is separated from other pairs with `,`, e.g., `'a':10, 'b':20`

````python
d = {'a':10, 'b':20, 'c':30}  # this dict has 3 name-value pairs  -- don't think about these as 6 items. They are 3!
````

````python
len(d)   # how many key-value pairs are in d?
````

````python
d['a']  # retrieve the value associated with 'a'
````

````python
d[a]   # this means: retrieve the value associated with the variable a's value!
````

# Rules for creating dicts

1. A dict can contain any number of key-value pairs. The minimum is zero (the empty dict, `{}`), and the maximum is determined by the memory in your computer.
2. Any immutable value in Python (i.e., basically numbers, strings, and tuples) can be a dict key. Normally, we use numbers and strings as keys. (This rule is slightly wrong, but not enough to cause trouble for you at this point.)
3. In any given dict, the keys must be unique
4. Every key has a value, and every value has a key.
5. There are NO RESTRICTIONS WHATSOEVER on values. They can be mutable. They can repeat. They can be big, nested, combinations of things.

You have to be very careful with dict keys -- they only return a value if there is a 100% match. Capitalization, whitespace, punctuation... all of those are crucial.

# Are dicts mutable?

We've already discussed that strings and tuples are immutable -- they cannot be changed, once we have defined them. Lists, of course, are mutable -- we can modify existing values, add new values, and remove existing ones.

What about dictionaries?

They are *mutable*. 

- We can replace an existing value
- We can add a new key-value pair
- We can remove an existing key-value pair

````python
d = {'a':10, 'b':20, 'c':30}

# How do I replace a value in this dict?
# answer: just assign with []!

d['a'] = 999
````

````python
d
````

````python
# How do I add a new key-value pair?
# Same way as we update a value -- just assign to the key

# If the key already exists, it's an update
# If the key doesn't yet exist, it's an addition to the dict

d['xyz'] = 987
d
````

````python
# How do I remove a key-value pair?
# we invoke dict.pop() on the key.
# The key-value pair is removed *and* the value is returned to the caller.

d.pop('xyz')
````

# Are keys kept in sorted order?

Sort of.

For a very long time, dicts were not stored in any predictable or guaranteed order. What came first vs. second when you iterated over it was a matter of a lot of luck.

In Python 3.6, they changed how dicts are implemented, and one side effect of that was that dicts are kept in *chronological* order. That is, the first key-value pair assigned is first, the second is second, etc.

# Exercise: Vowels, digits, and others (dict edition)

1. Define a dict, `counts`, with three keys -- `vowels`, `digits`, and `others`. All three values for these keys should be 0.
2. Ask the user to enter a string.
3. Go through the string, one character at a time:
    - If it's a vowel, add 1 to the value for `vowels`
    - If it's a digit, add 1 to the value for `digits`
    - If it's neither, add 1 to the value for `others`
4. Print `counts` in the end.

````python
counts = {'vowels':0,
          'digits':0,
          'others':0}

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in 'aeiou':   # if one_character is a vowel...
        counts['vowels'] += 1      # ... then add 1 to vowels
    elif one_character.isdigit():  # if one_character is a digit...
        counts['digits'] += 1      # ... then add 1 to digits
    else:
        counts['others'] += 1      # otherwise, just add to others

print(counts)        
````

# Using `[]` vs. others

It's true that we use different kinds of parentheses to create data structures:

- `[]` to create a list
- `()` to create a tuple
- `{}` to create a dict

However, in all of these cases, as well as in strings, we use `[]` to retrieve an element. Why? 

The answer is that when I say `x[i]` in Python, that is translated behind the scenes to what's known as a "magic method," one that we don't run ourselves, but that is run our behalf. In this case, the method is known as `__getitem__` ("dunder getitem"). So our invocation of `x[i]` becomes

    x.__getitem__(i)

Any object that defines `__getitem__` then can use `[]`. In the case of these core data structures, they support `[]` for that reason.

# What can we store in a dict?

Yes, the keys in a dict are important -- they are the way that we get data in and out. 

But because the values can be absolutely anything at all in Python, we can create complex data structures that combine the core ones. You can store integer and strings as a dict's values. You can also store lists. Or dicts. A dict of dicts is very useful -- the outer key is used to retrieve the record, and the inner key is used to retrieve a field in that record.

- If you have a list inside of a dict, just say `d['x'].append(1)`, and you've added 1 to the list that is stored for the `x` key.
- If you have a dict inside of a list, just say `d['x']['y'] += 5`, and you've added 5 to the dict key `y` within the dict key `x`.



# Exercise: Vowels, digits, and others (dict-list edition)

1. Define a dict, `counts`, with three keys -- `vowels`, `digits`, and `others`. The values for all three keys should be `[]`, the empty list.
2. Ask the user to enter a string.
3. Go through that string, one character at a time.
    - If it's a vowel, append that character to the `vowels` list
    - If it's a digit, append the digit to the `digits` list
    - Otherwise, append the character to the `others` list
4. Print `counts` in the end. You will see a dict of lists.

````python
counts = {'vowels':[],
          'digits':[],
          'others':[]}

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in 'aeiou':                 # if one_character is a vowel...
        counts['vowels'].append(one_character)   # ... then add 1 to vowels
    elif one_character.isdigit():                # if one_character is a digit...
        counts['digits'].append(one_character)   # ... then add 1 to digits
    else:
        counts['others'].append(one_character)   # otherwise, just add to others

print(counts)        
````

# Three paradigms of dict usage

There are a lot of ways to use a dict. But in many cases, they fit into three different paradigms. We've already seen two:

1. Define a dict, and never change it. Treat it like a read-only database.
2. Define a dict with keys and initial values (0, empty list, etc). The keys won't change, and we won't add/remove keys from the dict. But the values will be changed, and will be updated to reflect what the program has seen/computed. It's a way to accumulate over time using one variable rather than a bunch.
3. Define an empty dict, and then count the unknown -- that is, we don't know what we'll get or see, but we do know that we want to count whatever comes our way. Creating a dict with all possible keys is just too time- or space-consuming.

````python
# count characters in text
# I want to ask the user to enter some text, and count how many times each character appears.

counts = {'a':0, 'b':0, 'c':0, 'd':0, 'e':0}

text = input('Enter text: ').strip()

for one_character in text:
    counts[one_character] += 1

print(counts)    
````

Defining the dict with all possible characters as keys is just not realistic in this case. We'll do it in a different way:

1. The first time we see a character, we add a new key-value pair to our dict
2. The next time we see it, we add 1 to the value



````python
# version 2 of our character counter

counts = {}

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in counts:     # have we seen this character before?
        counts[one_character] += 1  # ... if so, add 1 to the count
    else:
        counts[one_character] = 1   # if it's new, then add a new key-value pair with the character and 1.

print(counts)        
````

# Exercise: Rainfall

The goal of this exercise is to keep track of the rainfall (in mm) in different cities over time. We don't know what cities we'll be measuring. We'll rely on the user to enter the name of a city, and then the amount of rain that fell there today. The dict keys will be strings (city names). The values will be integers (total mm rain in a city).

1. Define an empty dict, `rainfall`.
2. Ask the user to repeatedly enter the name of a city.
    - If the user enters an empty city name, break from the loop.
3. If you got a city name, then ask the user to enter mm rain.
    - Here, you can assume you'll get digits from the user
4. If you have seen data from this city before, add it to the existing number.
5. If you have not seen data from this city before, add a new key-value pair.
6. When the user finishes entering data, print the entire `rainfall` dict.

````python
rainfall = {}

while True:
    city_name = input('City: ').strip()

    if city_name == '':
        break

    mm_rain = input('Rainfall: ').strip()
    mm_rain = int(mm_rain)   # big assumption -- that we will get only integers from the user

    if city_name in rainfall:           # have we seen this city before?
        rainfall[city_name] += mm_rain  # ... if so, then just add mm_rain to the existing total
    else:
        rainfall[city_name] = mm_rain   # haven't seen this city before, so just set the value

print(rainfall)         
````

# How are dictionaries implemented?

The simplest way to understand a dict is sort of like a two-column table, like we saw in the Python Tutor display. The first column is keys and the second column is values. We ask for a value based on a key, and we get it. We store a value based on a key, and it's stored.

But that makes it look like the key-value pairs are stored in order, or in something we can track easily.

The more complex story is that key-value pairs are stored in Python not in order as we saw, but rather based on the result of running the `hash` function on the key.

That is: When we store a key-value pair, we store it in a location based on the output from `hash(key)`. 

What is `hash`? It's a function that returns an integer that *looks* random, but isn't. The idea is that this spreads the key-value pairs all over memory, and avoids the chance of a collision -- two different keys being stored in the same place.

Why go through all of this effort to store our key-value pairs?

Answer is: Super duper fast lookup and retrieval.

If you ask `'a' in d`, Python runs `hash('a')`, jumps to that location in memory, and can tell you right away if it's there.

The speed with which we can search in a dict's keys, or retrieve a value from a key, doesn't depend on the size of the dict. So whether it has 5 pairs or 5m pairs, Python will still find items in a dict very quickly.

Python loves dicts so much that it uses them to implement many of its features:

- Variables are strings -- and Python stores them in a dict, behind the scenes, with its values.
- Attributes, the names that come after `.`, are stored in a dict

So what problems are there with dicts?

What is a file? 

You can think of a file as a stored version of data structures from your computer's memory. The data structure only lasts so long as the computer is on, and only on one computer. To share it with others, or have a copy around when the user's computer restarts, is useful and necessary.

What we can do is take the data structures in memory, and dump them directly to a special file. And then, we can read them from the special file back into memory, and it's as if we never left.

Have you seen this before? Yes -- with Exel, Word, PowerPoint, etc.

Python can work with files -- both for writing and reading. However, we're going to stick with text files in this course -- they're common and also easy to work with.

In real life, you probably will work with other formats. 

We want to work with a file. How can we connect to it?

We can't, at least not direclty -- it would violate all sorts of security on your computer and operating system. Rather, you have to go through the OS. 

1. You `open` a file, telling the OS you want to read from it
2. Then you can read its contents into memory
3. Then you can (if appropriate) use the content with your program

Whatever we read from a file comes as a text string.

````python
filename = 'passwd.txt'

f = open(filename)  # open the file - -what we get back is a "file object," a go-between for Python and the OS
print(f.read())   # this returns the entirety of the file as one string
````

Don't use `read` to read from a file!

When you invoke `read`, you get back a string. What happens if you open a 100 TB file? 

There is a better way to read from a file, and it involves the `for` loop:

````python
filename = 'passwd.txt'

f = open(filename)

for one_line in f:    # by definition, looping over f returns a line
    print(one_line.strip())
````
